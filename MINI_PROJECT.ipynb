***USING PANDAS***

```python
import pandas as pd
df={
    "months":["january","february","march","april","may","june","july","august","september","october","november","december"],
    "sales":[150000,210000,100000,240000,200000,120000,230000,170000,220000,260000,140000,180000]
}
data1=pd.DataFrame(df)
print(data1)
print("TOTAL SALES")
print(data1.sum())
print("AVERAGE SALES")
print(data1.describe())
print("BEST MONTH")
print(data1.max())
print("WORST MONTH")
print(data1.min())
```

***USING MATPLOTLIB***

*1.Line chart*

```python
import matplotlib.pyplot as plt
months=["january","february","march","april","may","june","july","august","september","october","november","december"]
sales=[150000,210000,100000,240000,200000,120000,230000,170000,220000,260000,140000,180000]
plt.title("MONTHLY SALES")
plt.xlabel("MONTHS")
plt.ylabel("sales")
plt.plot(months,sales,marker=".")
plt.xticks(rotation=45)
plt.show()

```

*2.Bar chart*

```python
import matplotlib.pyplot as plt
months=["january","february","march","april","may","june","july","august","september","october","november","december"]
sales=[150000,210000,100000,240000,200000,120000,230000,170000,220000,260000,140000,180000]
plt.title("MONTHLY SALES")
plt.xlabel("MONTHS")
plt.ylabel("sales")
plt.bar(months,sales)
plt.xticks(rotation=45)
plt.show()
```
