## ***SERIES***

# ***INDEXING***

```python
import pandas as pd
students=['anu','likki','lohi','usha','kussu']
age=[21,24,26,22,25]
s=1,2,3,4,5
a=6,7,8,9,10
data1=pd.Series(students,index=s)
data2=pd.Series(age,index=a)
print(data1)
print(data2)
```
