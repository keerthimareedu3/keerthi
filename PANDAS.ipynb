# ***CONCATE***

```python
import pandas as pd
t=[[212,343,454,565],[213,312,415,516]]
l=[[324,453,463,231],[124,434,342,124]]
k=["A","B","C","D"]
df=pd.DataFrame(t,columns=k)
df2=pd.DataFrame(l,columns=k)
df3=(df,df2)
df4=pd.concat(df3)
print(df)
print(df2)
print(df4)
```

```python
import pandas as pd
m=[[24,355,232,555,666]]
k=[[3243,546,433,245,3423]]
l=["A","B","C","D","E"]
n=["F","G","H","I","J"]
df1=pd.DataFrame(m,columns=l)
df2=pd.DataFrame(k,columns=n)
df3=(df1,df2)
df4=pd.concat(df3,ignore_index=True)
df5=pd.concat(df3,axis=1)
print(df1)
print(df2)
print(df4)
print(df5)

```

```python
import pandas as pd
data={
    "product":["samsung","iphone","motorola"],
    "status":["success","success","failed"],
    "cost":[10000,50000,15000],
    "purdate":['02-sep-2019','here data is missing','21-sep-2019']
    }
df=pd.DataFrame(data)
df['purdate']=pd.to_datetime(df['purdate'],errors="coerce")
print(df)
print()
print(df.dtypes)
```
