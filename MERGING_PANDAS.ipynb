```python
import pandas as pd
p={
    "ID":[11,22,33,44,55],
    "NAMES":["KEERTHI","HARSHI","KUSSU","MAHITHA","NIMMI"]
   }
q={
    "ID":[11,22,33,44,55],
    "NAMES":["KEERTHI","ANUSHA","KUSSU","MAHITHA","JYOTHI"],
    "BRANCH":["CSE","ECE","CIVIL","EEE","AI"]
   }
df1=pd.DataFrame(p)
df2=pd.DataFrame(q)
df3=pd.merge(df1,df2,on="NAMES",how="right")
print(df1)
print(df2)
print(df3)
```

```python
import pandas as pd
m={
    "SI_NO":[1,2,3,4,5],
    "ITEM_NAMES":["MOBILE","LAPTOP","FRIDGE","WASHMACHINE","DISHWASHER"]
   }
k={
    "SI.NO":[1,2,3,4,5],
    "ITEM_NAMES":["MOBILE","LAPTOP","FRIDGE","WASHMACHINE","DISHWASHER"],
    "PRICES":[25000,55000,75000,25000,35000]
}
df1=pd.DataFrame(m)
df2=pd.DataFrame(k)
df3=pd.merge(df1,df2,on="ITEM_NAMES",how="right")
print(df1)
print(df2)
print(df3)
```
