```python
import pandas as pd
d1={
    "ID":[1,2,3,4,5],
    "NAMES":["kavya","kussu","jyothi","mahi","srilatha"]
    }
d2={
    "ID":[5,4,3,2,1],
    "NAMES":["srilatha","jyothi","kussu","kavya","mahi"],
    "SUBJECTS":["JAVA","PYTHON","C++","C","DOT NET"]
     }
df1=pd.DataFrame(d1)
df2=pd.DataFrame(d2)
df3=pd.merge(df1,df2,on="NAMES",how="inner")
print(df1)
print(df2)
print(df3)
```

```python
import pandas as pd
d1={
    "ID":[11,22,33,44,55,66],
    "EMP_NAMES":["ANU","LIKKI","LOHI","SRI","HARSHI","KUSSU"]

   }
d2={
    "ID":[33,55,22,66,11,44],
    "EMP_SAL":[1242,3541,7463,2332,6433,4683]
   }
df1=pd.DataFrame(d1)
df2=pd.DataFrame(d2)
df3=pd.merge(df1,df2,on="ID",how="inner")
print(df1)
print(df2)
print(df3)
```

```python
import pandas as pd
df=pd.read_csv("STUDENTS.csv")
df1=pd.read_csv("STUDENTS.1.csv")
df2=pd.merge(df,df1,on="STUDENTS",how="left")
print(df)
print(df1)
print(df2)
```

```python
import pandas as pd
k={
    "NAMES":["MAHI","SIRI","APPU","KUSSU","HARSHI"],
    "COURSE":["PYTHON","JAVA","C++","C","DAT NET"]
   }
L={
    "NAMES":["KUSSU","MAHI","KEERTHI","HARSHI","APPU"],
    "ID":[1,2,3,4,5]
}
m1=pd.DataFrame(k)
m2=pd.DataFrame(L)
m3=pd.merge(m1,m2,on="NAMES",how="left")
print(m1)
print(m2)
print(m3)
```

```python
import pandas as pd
k={
    "emp_names":["keerthi","harshi","jyothi","mahitha","anusha"],
    "emp_sal":[13234,32421,35242,23143,53421],
    "emp_id":[11,22,33,44,55]
   }
l={
    "emp_id":[11,22,33,44,55],
    "emp_location":["HYD","USA","HYD","USA","HYD"]
  }
df1=pd.DataFrame(k)
df2=pd.DataFrame(l)
df3=pd.merge(df1,df2,on="emp_id",how="left")
print(df1)
print(df2)
print(df3)
```
