```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
df=pd.read_csv("poly_dataset.csv")
print(df)
a=df.iloc[:,1:2].values
b=df.iloc[:,2].values
model=PolynomialFeatures(degree=6)
model.fit_transform(a,b)
obj=LinearRegression()
obj.fit(a,b)
plt.scatter(a,b)
plt.plot(a,obj.predict(a))
plt.show()
```

```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
df=pd.read_csv("poly_dataset.csv")
print(df)
a=df.iloc[:,1:2].values
b=df.iloc[:,2].values
model=PolynomialFeatures(degree=6)
x=model.fit_transform(a)
obj=LinearRegression()
obj.fit(x,b)
plt.scatter(a,b,color="red")
plt.plot(a,obj.predict(x),"yellow")
plt.show()
```
