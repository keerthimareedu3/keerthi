# 1.WEATHER DATA ANALYSIS


```python
import numpy as np
temperature= np.array([
    28, 30, 29, 31, 32, 33, 35, 36, 34, 33,
    32, 31, 30, 29, 28, 27, 26, 25, 24, 23,
    22, 21, 20, 22, 24, 26, 28, 30, 31, 33
])
print("HOTTEST DAY")
print(temperature.max())
print("COOLEST DAY")
print(temperature.min())
print("AVERAGE")
print(temperature.mean())
print("NO.OF DAYS")
print(temperature.shape)
modified_temperatures = np.where(
    temperature > 35, 35,
    np.where(temperature < 22, 22, temperature)
)

print("Modified Temperature Data:")
print(modified_temperatures)




```
