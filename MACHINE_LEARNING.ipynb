```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
d={
    "names":["KEERTHI","HARSHI","NAVYA","KAVYA","SRILATHA"],
    "location":["HYD","KHMM","KDD","HYD","KDD"]
  }
df=pd.DataFrame(d)
print("NOMRAL DATAFRAME",)
print(df)
label_encoder=LabelEncoder()
df["NAMES"]=label_encoder.fit_transform(df['names'])
df["LOCATION"]=label_encoder.fit_transform(df["location"])
print()
print("TRANSFORM DATAFRAME",)
print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
k={
    "items":["mobile","laptop","washmachine","fridge"],
    "brand_names":["apple","HP","samsung","haier"],
    "prices":[1000000,50000,25000,30000]
}
df=pd.DataFrame(k)
print("NORMAL DATAFRAME")
print(df)
data=LabelEncoder()
df["ITEMS"]=data.fit_transform(df["items"])
df["BRAND_NAMES"]=data.fit_transform(df["brand_names"])
df["PRICES"]=data.fit_transform(df["prices"])
print()
print("TRANSFORM DATAFRAME")
print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
df=pd.read_csv("sales31.csv")
sales=LabelEncoder()
df["ORDER_ID"]=sales.fit_transform(df["ord id"])
df["CUSTOMER_NAME"]=sales.fit_transform(df["cust name"])
df["CUSTOMER_ID"]=sales.fit_transform(df["cust id"])
df["PRODUCT_NAME"]=sales.fit_transform(df["prod name"])
df["PRODUCT_COST"]=sales.fit_transform(df["prod cost"])
print(df)
```
