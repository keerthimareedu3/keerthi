# **FIT_TRANSFORM(p)**

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
d={
    "names":["KEERTHI","HARSHI","NAVYA","KAVYA","SRILATHA"],
    "location":["HYD","KHMM","KDD","HYD","KDD"]
  }
df=pd.DataFrame(d)
print("NOMRAL DATAFRAME",)
print(df)
label_encoder=LabelEncoder()
df["NAMES"]=label_encoder.fit_transform(df['names'])
df["LOCATION"]=label_encoder.fit_transform(df["location"])
print()
print("TRANSFORM DATAFRAME",)
print(df)
```

# **MinMaxScaler**

# **FUNCTIONTRANSFORMER**

```python
import numpy as np
from sklearn.preprocessing import FunctionTransformer

a = [[10, 20], [30, 40], [50, 60]]

f = np.array(a)

def add_ten(x):
 return x + 10

obj = FunctionTransformer(add_ten)

result = obj.transform(f)

print(f)
print()
print(result)
```

# **MISSING VALUES**

```python
import numpy as np
from sklearn.impute import SimpleImputer
import pandas as pd
students = [
                   [85, 'M', 'verygood'],
                   [95, 'F', 'excellent'],
                   [60, None, 'good'],
                   [np.nan, 'M', 'average'],
                   [70, 'M', 'good'],
                   [np.nan, None, 'verygood'],
                   [60, 'F', 'verygood'],
                   [98, 'M', 'excellent']
]

cols = ['marks', 'gender', 'result']
df = pd.DataFrame(students, columns = cols)
print(df)
obj=SimpleImputer(missing_values=np.nan,strategy="mean")
result=df["marks"].values.reshape(-1,1)
df.marks=obj.fit_transform(result)
print()
print(df)
```

```python
import numpy as np
from sklearn.impute import SimpleImputer
import pandas as pd
k=[
    ['keerthi','1','CSE'],
    ['kavya','2','EEE'],
    ['harshi',np.nan,'ECE'],
    ['navya','4',' civil']
]
df=pd.DataFrame(k,columns=["names","id","branch"])
print(df)
obj=SimpleImputer(missing_values=np.nan,strategy='median')
data=df["id"].values.reshape(-1,1)
df.id=obj.fit_transform(data)
print()
print(df)
```

```python
import numpy as np
from sklearn.impute import SimpleImputer
import pandas as pd
k=[
    ['keerthi',234],
    ['harshi',243],
    ['kavya',np.nan],
    ['navya',345],
    ['kussu',np.nan],
    ['nimmi',324]
]
df=pd.DataFrame(k,columns=["students","score"])

print(df)
value=SimpleImputer(missing_values=np.nan,strategy="most_frequent")
data=df["score"].values.reshape(-1,1)
df.score=value.fit_transform(data)
print()
print(df)
```

# **IMPUTING MISSING VALUES**

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

students = [
                        [85, 'M', 'verygood'],
                        [95, 'F', 'excellent'],
                        [75, np.nan, 'good'],
                        [np.nan, 'M', 'average'],
                        [70, 'M', 'good'],
                        [np.nan, np.nan, 'verygood'],
                        [92, 'F', 'verygood'],
                        [98, 'M', 'excellent']
]

cols = ['marks', 'gender', 'result']
df = pd.DataFrame(students, columns = cols)

print(df)

imputer = SimpleImputer(missing_values = np.nan,
strategy='most_frequent')

result = df['gender'].values.reshape(-1, 1)

df.gender = imputer.fit_transform(result).ravel()

print()
print(df)



```

# **TRAIN_TEST_SPLIT**

```python
import numpy as np
from sklearn.model_selection import train_test_split
values=np.arange(10)
train,test = train_test_split(values,train_size=0.5,test_size=0.5)
print("Given data:",values)
print("Train data:", train)
print("Test data:",test)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
data=np.arange(100)
train,test=train_test_split(data,train_size=0.7,test_size=0.3,random_state=1)
print("Given data:")
print(data)
print("Trained data:")
print(train)
print("Tested data:")
print(test)
plt.title("THE GIVEN DATA")
plt.xlabel("TRAINED DATA")
plt.ylabel("TESTED DATA")

plt.plot(train, label="Train Data")
plt.plot(test, label="Test Data")

plt.legend()
plt.show()
```
