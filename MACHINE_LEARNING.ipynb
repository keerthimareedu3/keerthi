# **FIT_TRANSFORM(p)**

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
d={
    "names":["KEERTHI","HARSHI","NAVYA","KAVYA","SRILATHA"],
    "location":["HYD","KHMM","KDD","HYD","KDD"]
  }
df=pd.DataFrame(d)
print("NOMRAL DATAFRAME",)
print(df)
label_encoder=LabelEncoder()
df["NAMES"]=label_encoder.fit_transform(df['names'])
df["LOCATION"]=label_encoder.fit_transform(df["location"])
print()
print("TRANSFORM DATAFRAME",)
print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
k={
    "items":["mobile","laptop","washmachine","fridge"],
    "brand_names":["apple","HP","samsung","haier"],
    "prices":[1000000,50000,25000,30000]
}
df=pd.DataFrame(k)
print("NORMAL DATAFRAME")
print(df)
data=LabelEncoder()
df["ITEMS"]=data.fit_transform(df["items"])
df["BRAND_NAMES"]=data.fit_transform(df["brand_names"])
df["PRICES"]=data.fit_transform(df["prices"])
print()
print("TRANSFORM DATAFRAME")
print(df)
```

# **MinMaxScaler**

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
d={
    "x":["keerthi","harshi","navya","kavya","kussu"],
    "y":[4,65,80,120,150]
   }
df=pd.DataFrame(d)
data=MinMaxScaler()
print(df)
values=data.fit_transform(df[["y"]])
print(values)
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

d = {
        "x" : [10, 20, 30, 40, 50],
        "y" : [25, 50, 75, 100, 125]
}

df = pd.DataFrame(d)
mm_scale = MinMaxScaler(feature_range = (0, 1))

print(df)

df[["x", "y"]] = mm_scale.fit_transform(df[["x", "y"]])
print(df)
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
h={
    "k":[25,50,75,100,125,150],
    "l":[50,100,150,200,250,300]
}
df=pd.DataFrame(h)
values=MinMaxScaler(feature_range = (0, 1))
print(df)
df[["k","l"]]=values.fit_transform(df[["k","l"]])
print(df)
```

# **FUNCTIONTRANSFORMER**

```python
import numpy as np
from sklearn.preprocessing import FunctionTransformer

a = [[10, 20], [30, 40], [50, 60]]

f = np.array(a)

def add_ten(x):
 return x + 10

obj = FunctionTransformer(add_ten)

result = obj.transform(f)

print(f)
print()
print(result)
```

```python
import numpy as np
from sklearn.preprocessing import FunctionTransformer
k=[[35,43],[45,56],[67,78]]
f=np.array(k)
def add_ten(x):
  return x+10
object=FunctionTransformer(add_ten)
result=object.transform(f)
print(f)
print()
print(result)
```

```python
import numpy as np
from sklearn.preprocessing import FunctionTransformer
k=[[23,65,76],[253,5453,342]]
l=np.array(k)
def sub(x):
  return x-10
values=FunctionTransformer(sub)
results=values.transform(l)
print("original matrix")
print(l)
print("transformer matrix")
print(results)
```

```python
import numpy as np
from sklearn.preprocessing import FunctionTransformer
o=[[325465,23546578,243546578,2435678,24332546]]
p=np.array(o)
def div(x):
  return x/100
data=FunctionTransformer(div)
values=data.transform(p)
print("original matrix")
print(o)
print("transformer matrix")
print(values)
```
