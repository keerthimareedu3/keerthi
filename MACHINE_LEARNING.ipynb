# **IMPUTING MISSING VALUES**

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

students = [
                        [85, 'M', 'verygood'],
                        [95, 'F', 'excellent'],
                        [75, np.nan, 'good'],
                        [np.nan, 'M', 'average'],
                        [70, 'M', 'good'],
                        [np.nan, np.nan, 'verygood'],
                        [92, 'F', 'verygood'],
                        [98, 'M', 'excellent']
]

cols = ['marks', 'gender', 'result']
df = pd.DataFrame(students, columns = cols)

print(df)

imputer = SimpleImputer(missing_values = np.nan,
strategy='most_frequent')

result = df['gender'].values.reshape(-1, 1)

df.gender = imputer.fit_transform(result).ravel()

print()
print(df)



```

**Support Vector Machine**

```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
data={
    "x":[35,67,12,76,88,41,59],
    "y":[0,1,0,1,1,0,1]
}
df=pd.DataFrame(data)
x=df.drop("y",axis="columns")
y=df["y"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=90)
obj=SVC(kernel="linear")
obj.fit(x_train,y_train)
print(obj.predict([[95]]))
w = obj.coef_[0]
b = obj.intercept_[0]
plt.title("SUPPORT VECTOR MACHINE")
plt.xlabel("MARKS")
plt.ylabel("PASS OR FAIL")
plt.scatter(df.x,df.y)
decision_boundary = -b / w[0]
plt.axvline(x=decision_boundary, linestyle='-')
plt.show()
```

```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
df=pd.read_csv("income.csv")
print(df)
scaler=MinMaxScaler()
scaler.fit(df[["Income($)"]])
df["Income($)"]=scaler.transform(df[["Income($)"]])
scaler.fit(df[["Age"]])
df["Age"]=scaler.transform(df[["Age"]])
km=KMeans(n_clusters=3)
prediction=km.fit_predict(df[["Income($)","Age"]])
print(prediction)
df["Cluster"]=prediction
print(df)
df1 = df[df.Cluster==0]
df2 = df[df.Cluster==1]
df3 = df[df.Cluster==2]
print(df1)
print(df2)
print(df3)
plt.scatter(df1.Age,df1['Income($)'],color='green')
plt.scatter(df2.Age,df2['Income($)'],color='red')
plt.scatter(df3.Age,df3['Income($)'],color='black')

plt.scatter(km.cluster_centers_[:, 0], km.cluster_centers_[:, 1],
color='purple', marker='*',label='centroid')

plt.legend()

plt.show()
```

```python
import pandas as pd
from sklearn.cluster import KMeans
from matplotlib import pyplot as plt
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv("income.csv")

scaler = MinMaxScaler()

scaler.fit(df[['Income($)']])
df['Income($)'] = scaler.transform(df[['Income($)']])

scaler.fit(df[['Age']])
df['Age'] = scaler.transform(df[['Age']])

sse = []
k_rng = range(1,10)

for k in k_rng:
      km = KMeans(n_clusters=k)
      km.fit(df[['Age', 'Income($)']])
      sse.append(km.inertia_)

plt.xlabel('K')
plt.ylabel('Sum of squared error')
plt.plot(k_rng, sse)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

df=pd.read_csv("insurance_data.csv")
print(df)
X = df[["age", "bought_insurance"]]

obj=KMeans(n_clusters=2)
prediction=obj.fit_predict(X)
df["Cluster"]=prediction
print(df)
df1=df[df.Cluster==0]
df2=df[df.Cluster==1]
plt.scatter(df1.age,df1["bought_insurance"],color="blue")
plt.scatter(df2.age,df2["bought_insurance"],color="black")
plt.scatter(obj.cluster_centers_[:, 0],obj.cluster_centers_[:, 1],color='purple', marker='*',label='centroid')
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report,confusion_matrix
import warnings
warnings.filterwarnings("ignore")
iris=load_iris()
df=pd.DataFrame(iris.data,columns=iris.feature_names)
df["target"]=iris.target
df["flower_name"]=df.target.apply(lambda x:iris.target_names[x])
x=df.drop(["target","flower_name"],axis='columns')
y=df.target
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.4,random_state=200)
obj=KNeighborsClassifier(n_neighbors=5)
obj.fit(x_train,y_train)
k=obj.predict(x_test)
print(df)
print(k)
print(confusion_matrix(k,y_test))
print(classification_report(y_test,k))
print(obj.predict([[2.4,4.5,3.2,8.9]]))
plt.bar(df.target,df.flower_name)
plt.show()
```

# **Naives Bayes**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
import warnings
warnings.filterwarnings("ignore")
data={
    "study_hours":[12,5,8,3,9,2],
    "attendence":[50,65,43,76,99,34],
    "internal_marks":[65,43,24,75,98,54],
    "result":[76,34,24,77,93,59]
}
df=pd.DataFrame(data)
x=df.drop("result",axis="columns")
y=df["result"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3)
obj=GaussianNB()
obj.fit(x,y)
print(obj.predict(x_test))
print(obj.predict([[6,87,84]]))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
import matplotlib.pyplot as plt

value={
    "age":[12,45,61,24,31],
    "loan_amount":[15000,5000000,10000000,1500000,2500000],
    "loan":["rejected","Sanctioned","rejected","Sanctioned","Sanctioned"]
}
df=pd.DataFrame(value)
x=df.drop("loan",axis="columns")
y=df["loan"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=90)
model= MultinomialNB()
model.fit(x,y)
print(model.predict(x_test))
print(model.predict([[35,3000000]]))
print(model.predict([[16,10000000]]))
plt.title("NAIVE_BAYES")
plt.xlabel("LOAN")
plt.ylabel("AGE")
plt.scatter(df.loan,df.age,color="blue",marker=">")
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import BernoulliNB
import matplotlib.pyplot as plt

targets={
    "age":[67,8,34,78,28,13],
    "insurance":[1,0,1,0,1,0]
}
df=pd.DataFrame(targets)
x=df.drop("insurance",axis="columns")
y=df["insurance"]
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.7,random_state=15)
values=BernoulliNB()
values.fit(x,y)
print(values.predict(x_test))
print(values.predict([[59]]))
plt.title("NAIVE_BAYE")
plt.xlabel("INSURANCE")
plt.ylabel("AGE")
plt.plot(df.insurance,df.age,color="red")
plt.show()
```
