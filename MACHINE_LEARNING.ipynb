# **IMPUTING MISSING VALUES**

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

students = [
                        [85, 'M', 'verygood'],
                        [95, 'F', 'excellent'],
                        [75, np.nan, 'good'],
                        [np.nan, 'M', 'average'],
                        [70, 'M', 'good'],
                        [np.nan, np.nan, 'verygood'],
                        [92, 'F', 'verygood'],
                        [98, 'M', 'excellent']
]

cols = ['marks', 'gender', 'result']
df = pd.DataFrame(students, columns = cols)

print(df)

imputer = SimpleImputer(missing_values = np.nan,
strategy='most_frequent')

result = df['gender'].values.reshape(-1, 1)

df.gender = imputer.fit_transform(result).ravel()

print()
print(df)



```

```python
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
d={
    "Study_hours":[2,4,6,8,9,10],
    "score":[7.2,7.5,8.0,9.5,9.8,10]

}
df=pd.DataFrame(d)
x=df.Study_hours
y=df.score
slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
plt.title("R_VALUE")
plt.xlabel("study_hours")
plt.ylabel("score")
plt.scatter(x,y,marker="*")
print(df)
print(r_value)
print(p_value)
print(slope)
print(intercept)
print(std_err)
plt.show()
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
import warnings
warnings.filterwarnings("ignore")
df=pd.read_csv("salaries.csv")
input=df.drop('salary_more_then_100k',axis='columns')
target=df['salary_more_then_100k']
obj=LabelEncoder()
input["company_n"]=obj.fit_transform(input['company'])
input["degree_n"]=obj.fit_transform(input['degree'])
input["job_n"]=obj.fit_transform(input['job'])
inputs_n=input.drop(['company', 'job', 'degree'], axis='columns')
model=DecisionTreeClassifier()
model.fit(inputs_n,target)

print(df)
print(model.predict([[2,2,0]]))
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
data={
    "study_hour":[2,4,6,8,10],
    "attendence":[50,65,70,75,40],
    "internal_marks":[55,69,89,42,32],
    "result":["pass","pass","pass","fail","fail"]
}
df=pd.DataFrame(data)
input=df.drop("result",axis="columns")
target=df["result"]
model=LabelEncoder()
input["study_hour_n"]=model.fit_transform(input["study_hour"])
input["attendence_n"]=model.fit_transform(input["attendence"])
input["internal_marks_n"]=model.fit_transform(input["internal_marks"])
inputs_n=input.drop(['study_hour','attendence','internal_marks'],axis='columns')
obj=DecisionTreeClassifier()
obj.fit(inputs_n,target)
print(df)
print(obj.predict([[12,95,99]]))
```
