# **FIT_TRANSFORM()**----method in LabelEncoder

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
p={
    "s.no":[1,2,3,4,5,6],
    "course":["CSE","EEE","ECE","CIVIL","AI&MI","DS&CS"]
    }
df=pd.DataFrame(p)
print(df)
values=LabelEncoder()
df["SI.NO"]=values.fit_transform(df["s.no"])
df["COURSES"]=values.fit_transform(df["course"])
print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
k={
    "students_names":["keerthi","kavya","navya","harshi","kussu"],
    "city":["hyd","khmm","kdd","wrg","srpt"]
    }
df=pd.DataFrame(k)
print(df)
data=LabelEncoder()
df['STUDENTS_NAMES']=data.fit_transform(df["students_names"])
df["CITY_NAMES"]=data.fit_transform(df["city"])
print(df)
```

```python
import numpy as np
from sklearn.impute import SimpleImputer
import pandas as pd
data={
    "NAMES":["A","B","C","D"],
    "MARKS":[90,np.nan,75,np.nan]
}

df=pd.DataFrame(data)
obj=SimpleImputer(missing_values=np.nan,strategy="mean")
result=df["MARKS"].values.reshape(-1,1)
df["marks"]=obj.fit_transform(result)
df1 = df[df["MARKS"] >= 80]

print(df)
print(df1)

```
