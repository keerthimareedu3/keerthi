```python
import qrcode
from PIL import Image
from IPython.display import display
data="http//www.google.com"
qr=qrcode.QRCode(
    version=1,
    box_size=10,
    border=5
   )
qr.add_data(data)
qr.make(fit=True)
img=qr.make_image(fill="black",back_color="white")
display(img)
img.save("my_qrcode.png")
```

```python
import qrcode
from IPython.display import display
from PIL import Image

# Wedding invitation text
invitation_text = (
    "💍 Wedding Invitation 💍\n\n"
    "We cordially invite you to the wedding of\n"
    "Keerthi & saipriya\n\n"
    "📅 Date: 15 April 2026\n"
    "⏰ Time: 10:30 AM\n"
    "📍 Venue: Sri Lakshmi Convention Hall, Hyderabad\n\n"
    "Your presence is our happiness 🙏"
)

# Create QR Code
qr = qrcode.QRCode(
    version=1,
    error_correction=qrcode.constants.ERROR_CORRECT_Q,
    box_size=5,
    border=4
)

qr.add_data(invitation_text)
qr.make(fit=True)

# Create image
qr_img = qr.make_image(fill_color="black", back_color="white")

# Save file
qr_img.save("wedding_invitation_qr.png")

# Display QR Code
display(qr_img)
```
